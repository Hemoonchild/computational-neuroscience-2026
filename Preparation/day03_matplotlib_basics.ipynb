# Day 3 — Matplotlib Basics (object-oriented API)

Practicing the object-oriented matplotlib API (`fig, ax = plt.subplots()`) instead of the implicit pyplot state-machine style — single plots, subplot grids, multi-line styling, bar charts, histograms, and saving figures to disk.

**Status:** all 6 tasks completed.

```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

### Task 1 — Single line plot, object-oriented
Sine wave with axis labels, title, and legend.

```python
x = np.linspace(0,2*np.pi,100)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x,y,label='sin(x)')
ax.set_xlabel("x")
ax.set_ylabel('sin(x)')
ax.set_title('Sine and Cosine wave')
ax.legend()
plt.show()
```

### Task 2 — 2x2 grid of subplots
sin, cos, sin², cos² in a 2x2 grid, each with its own title.

```python
x = np.linspace(0,2*np.pi,100)
y_sin = np.sin(x)
y_cos = np.cos(x)
y_sin2 = np.sin(x)**2
y_cos2 = np.cos(x)**2

fig, axs = plt.subplots(2,2,figsize=(8,8))
axs[0,0].plot(x,y_sin)
axs[0,0].set_title("sin(x)")
axs[0,1].plot(x,y_cos)
axs[0,1].set_title("cos(x)")
axs[1,0].plot(x,y_sin2)
axs[1,0].set_title("sin^2(x)")
axs[1,1].plot(x,y_cos2)
axs[1,1].set_title("cos^2(x)")

plt.tight_layout()
plt.show()
```

### Task 3 — Two lines, one axes, styled differently
sin(x) and cos(x) on the same axes with distinct colors/linestyles and a legend.

```python
fig, ax = plt.subplots()

ax.plot(x,y_sin,label='sin(x)',color='pink',linestyle='-')
ax.plot(x,y_cos,label='cos(x)',color='gray',linestyle='--')

ax.legend()
plt.show()
```

### Task 4 — Bar chart (neuroscience-flavored)
Spike counts per neuron as a bar chart.

```python
neuron_ids = ['N1', 'N2', 'N3', 'N4', 'N5']
spike_counts = [12, 27, 8, 34, 19]

fig, ax = plt.subplots()
ax.bar(neuron_ids,spike_counts)
ax.set_ylabel('Spike Count')
ax.set_title('Spikes Per Neuron')

plt.show()
```

### Task 5 — Histogram
Simulated reaction times, histogram with 40 bins.

```python
reaction_times = np.random.normal(loc=300, scale=50, size=500)

fig, ax = plt.subplots()
ax.hist(reaction_times,bins=40)
ax.set_xlabel('Reaction Time(ms)')
ax.set_ylabel('Count')
ax.set_title('Distribution of Reaction Times')

plt.show()
```

### Task 6 — Saving a figure to disk
Re-saved the sine wave plot to a PNG and confirmed the file exists.

```python
x = np.linspace(0,2*np.pi,100)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x,y,label='sin(x)')
ax.set_xlabel("x")
ax.set_ylabel('sin(x)')
ax.set_title('Sine and Cosine wave')
ax.legend()

fig.savefig('Sine_Wave.png',dpi=150)
os.path.exists('Sine_Wave.png')
```
