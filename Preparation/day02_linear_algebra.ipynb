### Task 1 — Eigen decomposition & the defining equation
3x3 matrix. Compute eigenvalues/eigenvectors with `np.linalg.eig`. Check that `A @ v` is approximately `lambda * v` for one eigenpair.

```python
A = np.random.randint(10,size=(3,3))

eig = np.linalg.eig(A)
eig_vectors = eig.eigenvectors
eig_values = eig.eigenvalues

Av = A @ eig_vectors[:,0]
lambdav = eig_vectors[:,0]*eig_values[0]

print(f"lambda.v: {lambdav}\nA.v: {Av}")
```

### Task 2 — A transformation you already understand: rotation
90-degree 2D rotation matrix, applied to a couple of points.

```python
R = [[0,-1],[1,0]]
v1_before = [1,1]
v1_after = np.dot(R,v1_before)
print(f"vector: {v1_before} rotated: {v1_after}")

v2_before = [0,2]
v2_after = np.dot(R,v2_before)
print(f"vector: {v2_before} rotated: {v2_after}")
```

### Task 3 — Dot products & cosine similarity, from scratch
Two random length-10 vectors. Dot product and cosine similarity computed manually.

```python
vectors = np.random.randint(0,10,size=(2,10))
print(vectors)

dot_product = np.sum(vectors[0]*vectors[1])
print(f"dot product: {dot_product}")

cosine_similarity = dot_product / (np.sqrt(np.sum(vectors[0]**2)) * np.sqrt(np.sum(vectors[1]**2)))
print(f"Cosine Similarity: {cosine_similarity}")
```

### Task 4 — Solving a linear system
3x3 matrix `A`, vector `b`. Solved `Ax = b` with `np.linalg.solve`, verified with `np.allclose`.

```python
A = np.random.randint(10,size=(3,3))
b = np.random.rand(3)
x = np.linalg.solve(A,b)

print(f"A:\n {A} \nb: {b}")
print(f"x: {x}")
print("matches?",np.allclose(A@x,b))
```

### Task 5 — Symmetric matrices & orthogonal eigenvectors
Built a symmetric matrix (`A + A.T`), computed its eigenvectors, verified pairwise orthogonality (dot products near zero) — the property PCA relies on.

```python
A = np.random.randint(10,size=(3,3))
M = A + A.T

print("A:\n",A)
print("A.T:\n",A.T)
print("M:\n",M)

eigenvectors = np.linalg.eig(M).eigenvectors
print("v1 and v2 dot:",np.round(np.dot(eigenvectors[:,0],eigenvectors[:,1])))
print("v1 and v3 dot:",np.round(np.dot(eigenvectors[:,0],eigenvectors[:,2])))
print("v2 and v3 dot:",np.round(np.dot(eigenvectors[:,1],eigenvectors[:,2])))
```
