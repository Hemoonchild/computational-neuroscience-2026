# Day 4 — Matplotlib: Scientific Figures

Simulating a neuron's membrane potential as a random walk, visualizing a third variable through color, and building a shared-axis multi-panel figure.


```python
import numpy as np
import matplotlib.pyplot as plt
```

### Task 1 — Voltage trace with a threshold line
Membrane potential simulated as a random walk (cumulative sum of small random steps), offset near -70 mV, with a dashed threshold line at -55 mV.

```python
fluctuation = np.random.normal(loc=0.05,size=200)
membrane_potential = np.cumsum(fluctuation)

fig, ax = plt.subplots()
ax.plot(np.arange(0,200),membrane_potential-70)
ax.set_xlabel("Time(ms)")
ax.set_ylabel("Membrane potential(mV)")
ax.set_title('Simulated Voltage Trace')
ax.axhline(y=-55,linestyle='--',color='red',label='Threshold')
ax.legend()

plt.show()
```

### Task 2 — Scatter plot with a colorbar
50 points with a third variable (firing rate) mapped to point color, with a colorbar.

```python
points = np.random.randint(0,100,size=(2,50))
firing_rate = np.random.randn(50)*10

fig, ax = plt.subplots()
scatter = ax.scatter(points[0,:],points[1,:],c=firing_rate)
cbar = fig.colorbar(scatter,ax=ax)
cbar.set_label('Firing Rate(Hz)')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Scatter colored by firing rate')

plt.show()
```

### Task 3 — Two-panel figure, shared x-axis
Voltage trace and a derived spike train (threshold crossings), stacked with a shared x-axis.

```python
fig, axs = plt.subplots(2,1,sharex=True)
axs[0].plot(membrane_potential-70,label='Membrane Potential')
axs[0].axhline(-55,color='red',linestyle='--',label='Threshold')
axs[0].legend()
axs[0].set_ylabel('Voltage(mV)')

axs[1].plot(membrane_potential-70>-55,label='Spike Train')
axs[1].set_xlabel('Time Step(ms)')
axs[1].set_ylabel('Spike(0/1)')

plt.show()
```
