### Task 1 — Array creation & basic stats
Create a 1D numpy array of the first 20 even numbers, without a Python loop or list comprehension. Print shape, dtype, mean, and standard deviation.

```python
even = np.arange(2,42,2)
print(even)
print(even.shape)
print(even.dtype)
print(np.mean(even))
print(np.std(even))
```

### Task 2 — Indexing & slicing
5x5 random integer array. Extract the middle row, last column, a 2x2 corner subblock, and all values greater than 50 via boolean masking.

```python
from numpy import random

random_array = random.randint(0,100,(5,5))
print(random_array)

print("\nMiddle row: ",random_array[2,:])
print("Last column: ", random_array[:,-1])
print("Top-right 2x2 subblock: ", random_array[0:2,0:2])
print("Values greater than 50: ", random_array[random_array>50])
```

### Task 3 — Reshaping
1D array of 0-23, reshaped to (2,3,4), then two axes swapped.

```python
oneDarray = np.arange(0,24)
reshaped = oneDarray.reshape(2,3,4)
print(oneDarray)
print(reshaped)

swaped = np.swapaxes(reshaped,1,2)
print(swaped)

print(reshaped.shape, swaped.shape)
```

### Task 4 — Broadcasting
(5,3) array of 5 "neurons" x 3 "features." Subtract the per-feature (column) mean from every row without a loop.

```python
neurons = random.randint(0,5,size=(5,3))
print(neurons)

centered = neurons - neurons.mean(axis=0)
print("----------\n",centered)

```

### Task 5 — Vectorization vs. loops
Squared Euclidean distance between two length-100,000 vectors, computed via a manual loop vs. vectorized numpy, timed.

```python
import time

d = 100000
v1 = np.random.randn(d)
v2 = np.random.randn(d)

t1 = time.time()
dist = 0
for i in range(d):
    dist += (v1[i] - v2[i])**2
t2 = time.time()
print(f"manual for loop: {t2-t1}")

t3 = time.time()
dist = np.sum((v1-v2)**2)
t4 = time.time()
print(f"vectorized numpy: {t4-t3}")
```

### Task 6 — Randomness & convergence
1000 fair coin flips. Running (cumulative) mean after each flip, confirming convergence toward 0.5 — the law of large numbers.

```python
flips = random.randint(0,2,1000)
mean = np.cumsum(flips)/np.arange(1,1001)
print(mean[-10:])

```

### Task 7 — Spike train (neuroscience-flavored)
Binary spike train (length 100, spike probability 0.1). Spike count, spike indices, and inter-spike intervals.

```python
spike_train = random.binomial(n=1, p=0.1,size=100)
print(spike_train)

spike_num = np.sum(spike_train)
print(spike_num)

spike_indices = np.where(spike_train==1)[0]
print(spike_indices)

spike_intervals = np.diff(spike_indices)
print(spike_intervals)
```

### Task 8 — Trial-averaged activity
(10, 100) array of 10 trials x 100 timepoints, with a Gaussian bump around timepoint 50. Trial-averaged response and peak-detection via `np.argmax`.

```python
a = np.arange(100)
bump = np.exp(-(a-50)**2/2)
activity = random.randn(10,100) + 0.5 * bump

mean = activity.mean(axis=0)

peak = np.argmax(mean)
print(f"peak timepoint: {peak}")
```
